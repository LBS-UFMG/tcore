```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# load data
df = pd.read_csv("andrade.tsv", sep="\t")

# 10 variables
features = [
    "WBGT",
    "Speed",
    "Initial TCORE",
    "Body mass",
    "TCORE - TSKIN",
    "Mean TSKIN",
    "Sweat rate",
    "VO2MAX",
    "Heart rate",
    "Delta mass"
]

target = "End-TCORE "

# remove NaN (important!)
df = df[features + [target]].dropna()

X = df[features].values
y = df[target].values

# linear regression in-sample (OLS)
model = LinearRegression()
model.fit(X, y)

# prediction
y_pred = model.predict(X)

# R²
r2 = r2_score(y, y_pred)
print(f"R² = {r2:.4f}")
```
